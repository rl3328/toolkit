## Cis_association
### cis-QTL: nominal p-values for all variant-phenotype pairs
**Model Overview:**
This is a linear regression model
Y (dependent variable) = Gene expression level(RNA, protein, metabolite level .etc)
X (independent variables) = Genotype(Genotype dosage for a specific variant (0, 1, or 2)) + Covariates(e.g., age, sex, hidden factors)

**Regression Equation:**
y = β0 + β1genotype + β2covariate1 + β3*covariate2 + ... + ε

**Number of Pairs and Betas:**

Total pairs = Number of genes × Number of variants in cis-region
Number of β1 (slope) = Number of pairs

**Nonlinearity:**
if there's interaction, b_g is main slope, b_i is interaction slope, b_gi is the slope of the main*interaction term.
