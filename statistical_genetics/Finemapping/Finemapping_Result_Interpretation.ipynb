# Finemapping Results Interpretation

## susie_rss pipeline in `pecotmr`

### 1. Introduction:
- **Bayesian Variable Selection in Linear Regression (BVSR)**: L = L, max_iter = max_iter, both are adaptive.
- **Bayesian Conditional Regression (BCR)**: L = L, max_iter = 1, a special case of BVSR.
- **SER (Single Effect Regression)**: L = 1, max_iter = 1, a special case of BCR, assuming "one causal variant" assumption.

### 2. When to use
- **BVSR**: Most flexible, but might be subject to overfitting
- **BCR**: Preferred when we want to detect secondary signal (we suspect there's a secondary signal)
- **SER**: To some extent reliable to exclude the effect of LD, especially when BVSR is messy. 

### 3. Comparison:
- **Complexity**: single_effect < bayesian_conditional_regression < susie_rss
- **Computational cost**: single_effect < bayesian_conditional_regression < susie_rss
- **Ability to detect multiple effects**: single_effect < bayesian_conditional_regression ≤ susie_rss

## Underlying Mechanisms or Explanations for Credible Sets (CS) with Relatively High Correlation

1. **Strong Linkage Disequilibrium (LD)**:
   - One of the most common explanations for closely located CS.
   - Variants in close physical proximity are often in strong LD due to less frequent recombination.
   - Strong LD can make it difficult to distinguish the true causal variant.

2. **Co-regulation**:
   - Closely located CS might represent variants involved in the co-regulation of genes.
   - Could be enhancers, silencers, or other regulatory elements working together.

3. **Shared Regulatory Elements**:
   - CS might contain variants within the same regulatory element affecting multiple genes.

4. **Gene Clusters**:
   - In regions with gene clusters, CS might represent effects on different but related genes.

5. **Pleiotropy**:
   - Closely located CS might represent pleiotropic effects on multiple related traits.

6. **Complex Haplotype Structure**:
   - The region might have a complex haplotype structure leading to multiple apparent signals.

7. **Tagging of a Single Causal Variant**:
   - Both CS might be tagging the same underlying causal variant not directly genotyped or imputed.

8. **Allelic Heterogeneity**:
   - Multiple causal variants in close proximity affecting the same gene or biological pathway.

9. **Technical Artifacts**:
   - Could be due to imputation errors or mapping problems in complex genomic regions.

10. **Evolutionary History**:
    - Proximity might reflect recent duplications or insertions.

11. **Functional Domains**:
    - In coding regions, CS might represent variants affecting different nearby functional domains.

12. **Epistasis**:
    - CS might represent variants that interact epistatically.

13. **Chromatin Interactions**:
    - Proximity might reflect 3D chromatin structure.

14. **Alternative Splicing**:
    - CS might affect different splice variants of the same gene.
