#### multiple testing
Fine-mapping methods, including those that compute credible sets and posterior inclusion probabilities (PIP), generally do not involve multiple testing correction in the same way traditional genome-wide association studies (GWAS) do. Instead, they rely on Bayesian statistical frameworks to account for uncertainty and correlations among variants.

Why Multiple Testing Correction Is Not Required in Fine-Mapping:
* Bayesian Approach:
Fine-mapping uses posterior probabilities rather than p-values.
Bayesian models naturally incorporate prior information and account for linkage disequilibrium (LD) between variants, focusing on the probability of causality rather than controlling false positives across independent tests.

* Region-Specific Analysis:
Fine-mapping usually focuses on a specific genomic region (e.g., a locus identified from GWAS), not the entire genome.
Since the scope of inference is narrower, the need for stringent corrections like Bonferroni or FDR (false discovery rate) is reduced.

* Posterior Inclusion Probability (PIP):
PIP thresholds (e.g., 0.9) are used as evidence of causality for individual variants. This approach evaluates evidence for each variant within the context of the region, rather than across the genome, avoiding the need for multiple testing adjustments.
