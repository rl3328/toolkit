# Enrichment
Aim of enrichment: compare gene-sets and region with function instead of variants to improve interpretability
## Gene-Level Enrichment
These methods assess whether individual genes are enriched for specific biological functions, pathways, or disease associations.
1. **Overrepresentation Analysis (ORA)**
Counts genes in a predefined list and compares them to a background set.
**Excess of Overlap: similar to ORA, but focus on the overlaping part of two genes.**
Tools: DAVID, Enrichr, GOrilla, g:Profiler
2. **Gene Ontology (GO) Analysis**
Tests for enrichment in biological processes, molecular functions, and cellular components.
Tools: PANTHER, Gene Ontology Consortium tools
3. Tissue-Specific Enrichment
Identifies genes with expression enrichment in specific tissues.
Tools: GTEx, Human Protein Atlas
4. Disease/Phenotype Association
Maps genes to known disease or phenotype associations.
Tools: DisGeNET, Open Targets, MalaCards
5. Partitioned Heritability Enrichment
Tools: **LDSC (Linkage Disequilibrium Score Regression)**
## Gene-Set Level Enrichment
These approaches analyze groups of genes that share common biological functions.

1. **Gene Set Enrichment Analysis (GSEA)**
Ranks all genes based on their association with a phenotype and tests whether predefined gene sets are enriched.
Tools: GSEA (Broad Institute), fgsea (R package)
2. Pathway Enrichment
Identifies enriched pathways using curated databases.
Tools: **KEGG**, Reactome, WikiPathways
3. Network-Based Enrichment
Integrates protein-protein interaction (PPI) networks or co-expression networks.
Tools: STRING, GeneMANIA
d. Single-Cell RNA-Seq Enrichment
Identifies gene sets enriched in specific cell types or states.
Tools: Cell-type Enrichment (scMappR, scRNAseq databases)
## Variant-Level Enrichment
These methods test whether genetic variants (e.g., SNPs, indels) are enriched in specific functional categories.

1. GWAS Enrichment
Tests whether GWAS-identified variants are enriched in specific pathways or regulatory elements.
Tools: **MAGMA**, FUMA, **LDSC (Linkage Disequilibrium Score Regression)**
2. Regulatory Element Enrichment
Identifies variant enrichment in enhancers, promoters, or transcription factor binding sites.
Tools: GREGOR, H3K27ac/ATAC-seq-based methods
3. eQTL Enrichment
Tests whether variants are enriched in expression quantitative trait loci (eQTLs).
Tools: GTEx, SMR (Summary-data based Mendelian Randomization)
d. Chromatin-State Enrichment
Tests whether GWAS variants are enriched in epigenetically active regions.
Tools: ChromHMM, Roadmap Epigenomics

## Other Functional Enrichment Approaches
These integrate multiple types of biological data to assess enrichment.

1. Polygenic Risk Score (PRS) Enrichment
Tests whether a set of variants collectively explain disease risk.
Tools: PRSice, LDpred
2. Multi-Omics Enrichment
Integrates genomics, transcriptomics, and proteomics for functional enrichment.
Tools: MOFA, mixOmics
3. Evolutionary Constraint-Based Enrichment
Identifies genes under evolutionary constraint based on sequence conservation.
Tools: GERP++, PhyloP, pLI (from gnomAD)
d. Drug-Gene Enrichment
Tests whether disease-associated genes are targeted by existing drugs.
Tools: DrugBank, CLUE (Connectivity Map)




## Enrichment in eQTL Regions:

1. Affects gene expression levels
2. Related to transcriptional regulation
3. Located in various genomic regions
4. Influences DNA-protein interactions
5. Identified through population-based studies
6. Acts at the genomic level

## Enrichment in RNA Binding Sites:

1. Affects post-transcriptional processes
2. Found within RNA sequences
3. Influences RNA-protein interactions
4. Impacts mRNA processing, stability, localization
5. Identified through experimental or computational methods
6. Acts at the transcriptomic level

## Key Differences:

1. Stage of gene expression (transcriptional vs post-transcriptional)
2. Molecular targets (DNA vs RNA)
3. Scope of effect (overall expression vs specific RNA processes)
4. Level of action (genomic vs transcriptomic)

Both types of enrichment provide valuable, complementary information about gene regulation mechanisms.
