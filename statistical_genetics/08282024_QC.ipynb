## 1. Unique filtering & WASP correction
### WASP (Weighted Alignment to Suppress Polarization) 
It is a method used to reduce reference allele bias in RNA-seq data, especially when dealing with heterozygous variants. WASP uses a simple approach to overcome mapping bias that can be readily incorporated into any read mapping pipeline. First, reads are mapped normally using a mapping tool selected by the user; mapped reads that overlap single nucleotide polymorphisms (SNPs) are then identified. For each read that overlaps a SNP, its genotype is swapped with that of the other allele and it is re-mapped. If a re-mapped read fails to map to exactly the same location, it is discarded (Fig. 1b). Unknown polymorphisms in the sample are not considered but will typically have little effect since the tests of allelic imbalance are only performed at known heterozygous sites. We performed a simulation to assess the impact of unknown polymorphisms and found that the proportion of heterozygous sites with biased mapping is very small (https://www.ncbi.nlm.nih.gov/pmc/articles/PMC4626402/).
**The grep -v "vW:i:[2-7]" part is crucial for deminishing allele-specific bias.**
It removes reads with WASP flags 2 through 7, which typically indicate: 
* 2: Read maps to the same position after changing the allele
* 3: Read maps to a different position after changing the allele
* 4-7: Various cases where the read or its mate don't map uniquely after changing the allele
* By filtering out these reads, the function effectively keeps only uniquely mapping reads that are not affected by potential reference allele bias.

### Unique filtering

* Percent Splice In (PSI) quantifies the usage of a specific splice variant compared to the total splicing events at a given locus. PSI values range from 0 to 1 (or 0% to 100%), representing the proportion of reads supporting the inclusion of a particular exon or splice variant. Phenotype data are not needed for the calculation but can provide valuable context for interpreting PSI values. Integrating these data types allows researchers to explore associations between splicing patterns and observable traits or genetic variants.
