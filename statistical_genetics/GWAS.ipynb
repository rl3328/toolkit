Using **public GWAS summary statistics** requires careful consideration to ensure validity and interpretability of results. Here’s a checklist of key aspects you **should know or check** before using them:

---

### ✅ **1. Genomic Build / Reference Genome**

* Make sure the summary stats use the **same genome build** (e.g., GRCh37/hg19 or GRCh38/hg38) as your reference panel and annotation datasets.
* If not, use tools like **LiftOver** to convert.

---

### ✅ **2. SNP Identifiers and Positions**

* Summary stats should include:

  * `SNP ID` (e.g., rsID or chr\:pos)
  * `CHR`, `POS` (chromosome, base-pair position)
  * `Effect Allele (EA)` and `Non-effect Allele (NEA)`
* If rsIDs are missing or outdated, map them using dbSNP or a reference VCF.

---

### ✅ **3. Effect Size Format**

* Know what is reported:

  * `BETA`, `OR`, `log(OR)`, or `Z-score`
  * `SE` (standard error)
  * `P-value`
* Make sure effect size aligns with the **effect allele**.

---

### ✅ **4. Sample Size**

* Look for per-SNP sample size (`N`) or total sample size.
* Some tools (like FINEMAP, SuSiE, and TWAS) adjust for sample size in likelihood or LD matrix scaling.

---

### ✅ **5. Allele Frequency**

* `EAF` (effect allele frequency) is often used for:

  * Filtering rare variants
  * Harmonizing alleles with LD panels
* If missing, estimate from a matched reference panel.

---

### ✅ **6. QC Status and Filtering**

* Ensure SNPs were:

  * Properly imputed (e.g., INFO > 0.8)
  * Not strand ambiguous (A/T, C/G)
  * Well-aligned with reference panel (check strand flips or allele mismatches)
* Remove SNPs with:

  * Missing values
  * Low imputation quality
  * High heterogeneity (if meta-analyzed)

---

### ✅ **7. Ancestry and Cohort Information**

* Match the **ancestry** of GWAS with your **LD reference panel**.

  * Use European LD panels for European GWAS, etc.
  * Using mismatched ancestry can bias fine-mapping/TWAS.

---

### ✅ **8. LD Reference Panel**

* Required for **fine-mapping** and some TWAS methods.
* Should be:

  * Matched in ancestry
  * Imputed or genotyped similarly
  * Have the same variant set and allele alignment

---

### ✅ **9. Region-Specific Information**

* For **fine-mapping**, ensure you can extract:

  * All SNPs in a locus (±1Mb or as defined)
  * With accurate LD and effect sizes
* Missing nearby SNPs can harm fine-mapping resolution.

---

### ✅ **10. Traits and Units**

* Ensure you understand the phenotype:

  * Is it binary or continuous?
  * How is it measured or transformed?
* This affects interpretation in downstream methods (e.g., TWAS effect direction).

---

### ✅ **11. Licenses and Terms of Use**

* Some public GWAS data may have restrictions on commercial use, redistribution, or publication.
* Check the accompanying **README or documentation**.

---

If you're planning to use them in **TWAS**, also make sure:

* Expression prediction models (e.g., from GTEx, PrediXcan, or FUSION) are trained in **matching tissue and ancestry**.
* GWAS summary stats have enough overlapping SNPs with the TWAS models and LD reference panel.
