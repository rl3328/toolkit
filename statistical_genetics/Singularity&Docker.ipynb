# Introduction to Containers, Images, and Singularity

## What is a Container?
A **container** is a lightweight, standalone, and executable package that includes everything needed to run an application:  
- **Code**,  
- **Runtime**,  
- **Libraries**, and  
- **Dependencies**  

Containers ensure **consistency** across different environments (e.g., development, testing, and production). They are isolated from the host system but share the same OS kernel.

## What is an Image?
An **image** is a blueprint or template for creating containers. It is a static file containing all the necessary dependencies and configurations. A **container runs as an instance of an image**.

- Think of an **image** as a **recipe**, and a **container** as a **dish** prepared from that recipe.
- Images are **immutable** (cannot be changed) and can be **shared** or stored in repositories like **Docker Hub** or **Singularity Hub**.

## What is Singularity?
**Singularity** is a containerization platform designed for **high-performance computing (HPC)** and **research applications**. Unlike **Docker**, which requires root privileges, **Singularity** is designed for security and allows users to run containers **without root access**, making it ideal for use on **HPC clusters**.

### Why Use Singularity?
- **Security**: Runs as the user (no root privileges required).
- **HPC Compatibility**: Works well on supercomputers and clusters.
- **Single-file Format**: Singularity images are contained in a `.sif` file, making them portable.

---

## How Do They Work?

### 1. Building an Image
In **Singularity**, you create an image file (`.sif` – **Singularity Image Format**). You can convert **Docker images** to Singularity images using:

```sh
singularity build my_image.sif docker://ubuntu:latest
```

---

### 2. Running a Container
To execute a container using an image:

```sh
singularity run containers/rna_quantification.sif
```

This command **starts** a container using `rna_quantification.sif`.

---

### 3. Accessing a Container
- To **open an interactive shell** inside a container:
  ```sh
  singularity shell containers/rna_quantification.sif
  ```
- To **execute a command** inside the container:
  ```sh
  singularity exec containers/rna_quantification.sif some_command
  ```

---

### 4. Binding Directories
You can access **host system files** inside the container by binding directories:

```sh
singularity exec --bind /home/user/data:/mnt containers/rna_quantification.sif ls /mnt
```

This allows you to read and write data inside the container while keeping the data on your local system.

---

### 5. Running a Script Inside the Container
If you need to run a script inside the container, you can use:

```sh
singularity exec containers/rna_quantification.sif python /scripts/quantify_rna.py
```

Or for shell scripts:

```sh
singularity exec containers/rna_quantification.sif bash /scripts/run_analysis.sh
```

---

## Key Differences Between Docker and Singularity

| Feature         | Docker | Singularity |
|----------------|--------|------------|
| **User Privileges** | Requires root (by default) | Runs as the user (non-root) |
| **HPC Compatibility** | Not optimized for HPC | Designed for HPC environments |
| **Image Format** | Layered (`.tar`) | Single file (`.sif`) |
| **Security** | Needs extra setup for security | More secure by design |
| **Execution Model** | Daemon-based (background process) | No daemon, runs directly |

---

## Example: Running RNA Quantification in Singularity
If you have an RNA quantification pipeline packaged in a Singularity image (`rna_quantification.sif`), you can execute it as follows:

```sh
singularity run containers/rna_quantification.sif --input /data/sample.bam --output /results/
```

For interactive debugging:

```sh
singularity shell containers/rna_quantification.sif
```

To execute a script inside the container:

```sh
singularity exec containers/rna_quantification.sif python /scripts/quantify_rna.py
```

---

## Summary
- **Containers**: Portable environments for software execution.
- **Images**: Pre-built templates for containers.
- **Singularity**: A containerization tool optimized for HPC environments.
- **Usage**: `singularity run`, `singularity exec`, `singularity shell`.
- **Security**: Singularity runs as a non-root user, making it safe for HPC clusters.

Singularity is a powerful tool for research computing, enabling reproducibility and portability for scientific workflows. 🚀
