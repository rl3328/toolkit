## 1. what are those files used for?

* *Aligned.toTranscriptome.out.bam can be used for further transcript/expression abundance estimation with various transcript quantification software that require reads to be mapped to transcriptome, such as RSEM or eXpress. In short, STAR for alignment, RSEM for quantification.

## 2. what is the difference between Aligned.toTranscriptome.out.bam and Aligned.sortedByCoord.out.bam

* The differences mainly come into two part:
* 1. Content: *Aligned.sortedByCoord.out.bam contains the alignments of the RNA-seq reads to the reference genome, sorted by their genomic coordinates whereas *Aligned.toTranscriptome.out.bam contains the alignments of the RNA-seq reads to the reference transcriptome. Note, that STAR first aligns reads to entire genome, and only then searches for concordance between alignments and transcripts. This approach might offer certain advantages compared to the alignment to transcriptome only, by not forcing the alignments to annotated transcripts [from STAR manual](https://physiology.med.cornell.edu/faculty/skrabanek/lab/angsd/lecture_notes/STARmanual.pdf).
  3. Further Purposes: *Aligned.sortedByCoord.out.bam can be used for genome-guided analysis(e.g. identifying novel splice junctions).*Aligned.toTranscriptome.out.bam can be used for transcript-level analysis(e.g. RNA quantification).

## 3. Do Aligned.toTranscriptome.out.bam files need unique map? if so, is -q 255 the correct parameter for them?

* Usually unique mapping describes a read which mapped only once in a genome with [a given number of mismatches](https://www.seqanswers.com/forum/bioinformatics/bioinformatics-aa/1928-unique-mapped-reads-definition-confusing), not in transcriptome. While we can generate `*Aligned.sortedByCoord.out.bam` files with unique filtering, the biological interpretation of such files may be limited. Applying unique filtering to these files is not common practice for most downstream applications. Reasons are as follows:
    * a) Tool Compatibility: Many downstream analysis tools (e.g., RSEM for expression quantification) are designed to handle multi-mapped reads in transcriptome BAM files, often using probabilistic assignment methods.
    * b) Preserving Biological Information: Multiple transcript isoforms of the same gene or transcripts from the same gene family might align to the same genomic location. Strict unique filtering could lead to ambiguity and loss of important information, particularly for:
        - Isoform detection
        - Gene family analysis

## 4. MAPQ Scores in STAR

- STAR uses MAPQ 255 by default to denote uniquely mapped reads in the genome.
- This default can be adjusted using the `--outSAMmapqUnique` parameter (integer 0 to 255) to ensures compatibility with downstream tools like GATK, which often use [AMPQ 60](https://github.com/alexdobin/STAR/issues/1069) for unique mappers.
- MAPQ 255 is a conservative value to define unique mapped reads.There shows the definition of [MAPQ](https://www.acgt.me/blog/2014/12/16/understanding-mapq-scores-in-sam-files-does-37-42).
- It's important to note that MAPQ scores from different software cannot be directly compared due to variations in calculation methods.
