## 2. How to read rds files
* open R/Rstudio softare
* load the rds file using `data =readRDS("path/to/file.rds")`
* viewing rds:
* if using R studio, you can see the structure of each layers(the part before ":" is the name of that layer)
* if using base R, see the structure using `str(data)`

```python
Saving data into R data formats can reduce considerably the size of large files by compression.
# Save an object to a file
saveRDS(object, file = "my_data.rds")
# Restore the object
readRDS(file = "my_data.rds")
```
