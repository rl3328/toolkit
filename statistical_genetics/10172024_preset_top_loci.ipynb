## Issues During Merging of RDS Files

1. Upstream results for MetaBrain rss_eqtl and xQTL don’t fully overlap because MetaBrain eqtl_rss includes all regions, even non-transcript regions. Many non-productive regions don’t have rss_eqtl results (empty lists), so I’ve only combined the non-empty regions. Details can be found [here](https://github.com/rl3328/xqtl-analysis/discussions/9)..

2. I first merged the metabrain.tsv and xqtl.tsv files, removed rows with empty conditions, and then merged the RDS files based on this combined TSV file.

3. Issues that may arise when combining RDS files:

* The xqtl.rds path may not exist.
* The metabrain.rds path may not exist.
* metabrain.rds may be null (i.e., length(new) == 0). I’ve verified that no xQTL RDS files are empty (from the pipeline).
* Out of 9 regions with an RDS but no corresponding TSV, I was able to retrieve 5 TSV files. Of the remaining 4, one RDS is corrupted and three are empty.
* Some xqtl.rds files from an earlier, undeveloped cloud pipeline are corrupted and cannot be read.
